## TASK 2 — Data Visualization

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt # visualizing data
%matplotlib inline
import seaborn as sns
```

```python
df = pd.read_csv(r'C:\J_Notebook\CodeAlpha_ProjectName\Task_1\Titanic-Dataset.csv', encoding = 'unicode_escape')
```

```python
df.fillna({
    'Age':df['Age'].median,
    'Embarked':df['Embarked'].mode()[0]
    },inplace = True)
df.drop('Cabin', axis=1, inplace = True)
```

```python
# Survival Count Chart (Bar Chart)

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Survived",
palette=["#e74c3c", "#2ecc71"], hue = 'Survived', legend = True)
plt.title("Titanic — Survived vs Not Survived", fontsize=14)
plt.xlabel("Survival Rate")
plt.ylabel("No. Of Passengers")
plt.xticks([0, 1], ["Did Not Survive", "Survived"])
plt.tight_layout()
plt.savefig("survival_count.png", dpi=150)
plt.show()
```

```python
#  Gender vs Survival (Grouped Bar)

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Sex", hue="Survived",
palette=["#e74c3c", "#2ecc71"])
plt.title("Survival by Gender", fontsize=14)
plt.xlabel("Gender")
plt.ylabel("No. Of Passengers")
plt.legend(["Did Not Survive", "Survived"])
plt.tight_layout()
plt.savefig("gender_survival.png", dpi=150)
plt.show()
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```

```python
#  Age Distribution (Histogram)

plt.figure(figsize=(10, 5))
sns.histplot(data=df, x="Age", hue="Survived",
bins=30, kde=True,
palette=["#e74c3c", "#2ecc71"])
plt.title("Age Distribution — Survivors vs Non-Survivors",
fontsize=14)
plt.xlabel("Age")
plt.ylabel("Age Of Passengers")
plt.tight_layout()
plt.savefig("age_distribution.png", dpi=150)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x="Pclass", y="Fare",
palette="Set2", hue = "Pclass")
plt.title("Fare Distribution by Passenger Class", fontsize=14)
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.tight_layout()
plt.savefig("fare_boxplot.png", dpi=150)
plt.show()
```
